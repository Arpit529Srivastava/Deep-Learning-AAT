# Basic Tensor Operations using TensorFlow

**Objective:** To perform and understand basic tensor operations (addition, subtraction, multiplication, matrix multiplication, division, transpose, reshape, and reduction operations) using TensorFlow.

## 1. Theory

A **tensor** is a multi-dimensional array, the core data structure used by TensorFlow to represent data.

- **Scalar (0-D):** a single number
- **Vector (1-D):** a list of numbers
- **Matrix (2-D):** rows and columns
- **N-D Tensor:** used for images, batches of data, etc.

TensorFlow uses tensors instead of plain Python lists because tensors support GPU acceleration, automatic differentiation, and fast vectorized computation, all of which are essential for deep learning.

## 2. Setup

```python
import tensorflow as tf
print("TensorFlow version:", tf.__version__)
```

## 3. Creating Tensors

```python
# Scalar (0-D tensor)
scalar = tf.constant(7)
print("Scalar:\n", scalar)

# Vector (1-D tensor)
vector = tf.constant([1, 2, 3])
print("\nVector:\n", vector)

# Matrix (2-D tensor)
matrix = tf.constant([[1, 2], [3, 4]])
print("\nMatrix:\n", matrix)

# 3D Tensor
tensor3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("\n3D Tensor:\n", tensor3d)
```

```python
# Special tensors
zeros = tf.zeros([2, 3])
ones = tf.ones([3, 2])
random_tensor = tf.random.uniform([2, 2], minval=0, maxval=10)

print("Zeros:\n", zeros)
print("\nOnes:\n", ones)
print("\nRandom Tensor:\n", random_tensor)
```

## 4. Basic Tensor Operations

We'll use two 2x2 matrices `a` and `b` for the operations below.

```python
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

print("Tensor a:\n", a)
print("\nTensor b:\n", b)
```

### 4.1 Addition

```python
add_result = tf.add(a, b)
print("a + b =\n", add_result)
```

### 4.2 Subtraction

```python
sub_result = tf.subtract(a, b)
print("a - b =\n", sub_result)
```

### 4.3 Element-wise Multiplication

```python
mul_result = tf.multiply(a, b)
print("a * b (element-wise) =\n", mul_result)
```

### 4.4 Matrix Multiplication (Dot Product)

This is different from element-wise multiplication. `tf.matmul()` performs true matrix multiplication (rows x columns), which is what's used in neural network layers.

```python
matmul_result = tf.matmul(a, b)
print("a @ b (matrix multiplication) =\n", matmul_result)
```

### 4.5 Division

```python
div_result = tf.divide(a, b)
print("a / b =\n", div_result)
```

### 4.6 Transpose

```python
transpose_result = tf.transpose(a)
print("Transpose of a =\n", transpose_result)
```

### 4.7 Reshape

```python
reshape_result = tf.reshape(a, [4, 1])
print("Reshaped a (4x1) =\n", reshape_result)
```

### 4.8 Reduction Operations

```python
print("Sum of all elements in a:", tf.reduce_sum(a).numpy())
print("Mean of all elements in a:", tf.reduce_mean(a).numpy())
print("Max value in a:", tf.reduce_max(a).numpy())
print("Index of max value in each row of a:", tf.argmax(a, axis=1).numpy())
```

### 4.9 Element-wise Square

```python
square_result = tf.square(a)
print("Square of a (element-wise) =\n", square_result)
```

## 5. Constants vs Variables

`tf.constant` creates an immutable tensor, its value cannot change. `tf.Variable` creates a mutable tensor, used for things like weights that get updated during training.

```python
const_val = tf.constant(10)
var_val = tf.Variable(10)

var_val.assign(20)  # allowed
print("Constant:", const_val.numpy())
print("Updated Variable:", var_val.numpy())

# const_val.assign(20)  # This would raise an error, constants can't be reassigned
```

## 6. Observations

- `tf.add`, `tf.subtract`, `tf.multiply`, and `tf.divide` all operate **element-wise**.
- `tf.matmul` performs true matrix multiplication and requires compatible dimensions (columns of first = rows of second).
- `tf.reshape` changes the shape of a tensor without changing its data, as long as the total number of elements stays the same.
- `tf.Variable` is mutable and is the data type used for trainable parameters in neural networks, while `tf.constant` is fixed.
